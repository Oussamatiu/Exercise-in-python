```python
from math import sqrt , inf



def charger_villes(chemin):
   with open(chemin , "r") as fichier :
       lines = fichier.readlines()
       global villes
       villes = tuple(lines)
    #    print(len(villes) , villes)
       return villes
charger_villes("villes.txt")

def distance(villeA , villeB):
   
  
    xA = float(villeA[-1])
    yA = float(villeA[-2])

    xB = float(villeB[-1])
    yB = float(villeB[-2])

    return sqrt((xB - xA)**2 + (yB - yA)**2)


for x in range(2):
    print(distance( tuple(villes[x].strip("\n").split()) , tuple(villes[x + 1].strip("\n").split())))
   


def itineraire_greedy(villes):
    
    ville = villes[10]
    villees = []
    villees.append(ville)
    while len(villees) < len(villes) :
        ville_proche = min(
        (x for x in villes if x not in villees),
        key=lambda x: distance(
            tuple(ville.strip("\n").split()),
            tuple(x.strip("\n").split())
        )
        )
        ville = ville_proche
        villees.append(ville)
    return villees    
   

itineraire = itineraire_greedy(villes)

print(itineraire)

def distance_totale(itineraire):
    return distance(tuple(itineraire[1].strip("\n").split()) , tuple(itineraire[-1].strip("\n").split()))

distance_totale(itineraire) 

print("Nombre total de villes :", len(villes))
print("Itinéraire trouvé :")

for ville in itineraire:
    print(ville.strip("\n"))

print("Distance totale :", distance_totale(itineraire))


   
```
