## Exercice 1 : Classe Complète avec POO

Créez un système de gestion de véhicules en utilisant les concepts de POO vus dans ce cours.

### Consignes :

1. **Classe de base `Vehicule`** :
   - Attributs : `marque` (protégé), `vitesse_max` (privé)
   - Méthodes : `__init__`, `__str__`, `__eq__`, `deplacer()`

2. **Classes dérivées** :
   - `Voiture` : hérite de `Vehicule` et ajoute l'attribut `portes`
   - `Moto` : hérite de `Vehicule`

3. **Comportements attendus** :
   - `deplacer()` doit retourner un message spécifique pour chaque type de véhicule
   - `__eq__` doit comparer deux véhicules selon leur marque et leur vitesse maximale
   - `__str__` doit retourner une représentation lisible du véhicule

4. **Test final** :
   - Créez une liste contenant différents types de véhicules
   - Parcourez la liste et utilisez le polymorphisme pour appeler `deplacer()` sur chaque véhicule
   - Testez également l'égalité entre deux véhicules

````python
from abc import ABC , abstractclassmethod
class Vehicule(ABC) :
    def __init__(self , marque ,vitesse_max):
         self._marque = marque
         self.__vitesse_max = vitesse_max
    
    def get_vitesse_max(self):
        return self.__vitesse_max

    def __str__(self):
         return f"le marque de Vehicule est {self._marque} et le max vitesse est {self.get_vitesse_max()}"
    def __eq__(self , Vehicule):
         return self._marque == Vehicule._marque and self.__vitesse_max == Vehicule.get_vitesse_max()
    @abstractclassmethod    
    def deplacer(self):
        pass




class Voiture(Vehicule):
    def __init__(self, marque, vitesse_max, portes):
        super().__init__(marque, vitesse_max)
        self.portes = portes

    def deplacer(self):
        return "Voiture"

class Moto(Vehicule):
    def __init__(self, marque, vitesse_max):
        super().__init__(marque, vitesse_max)
        
    def deplacer(self):
        return "Moto"

list =  [Voiture("BMW" ,336 , 2), Moto("BMW",90) , Voiture("Audi" , 300 , 4)]

for x in list :
    print(x.deplacer())
print(list[0] == list[2])
````
