```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

clients = pd.read_csv("dataset/clients_ml.csv")

print("Informations générales : ")
print(f"Nombre de dimensions : {clients.ndim}")
print("\nTypes des colonnes :")
print(clients.dtypes)

print("\nValeurs manquantes : ")
print(clients[clients.isna().any(axis=1)])

print("\nDoublons : ")
print(clients[clients.duplicated()])

print("\nNettoyage de l'âge :")

moyenne_age = clients["age"].mean()
print(f"Âge moyen : {moyenne_age:.2f}")
clients["age"] = clients["age"].fillna(moyenne_age)
clients["age"] = clients["age"].astype(int)
clients = clients.drop_duplicates()

print("Valeurs manquantes de l'âge remplacées.")
print("Doublons supprimés.")

print("\nTypes après nettoyage :")
print(clients.dtypes)


print("\nNettoyage du salaire : ")

clients["salaire"] = (
    clients["salaire"]
    .astype(str)
    .str.replace(" DH", "", regex=False)
)

clients["ville"] = clients["ville"].str.strip().str.lower()

clients["salaire"] = pd.to_numeric(
    clients["salaire"],
    errors="coerce"
)

print("Salaire converti en valeur numérique.")
print("Noms des villes standardisés.")


print("\n===== Détection des valeurs aberrantes =====")

Q1 = clients["salaire"].quantile(0.25)

Q3 = clients["salaire"].quantile(0.75)

IQR = Q3 - Q1

limite_inf = Q1 - 1.5 * IQR

limite_sup = Q3 + 1.5 * IQR

print(f"Q1 : {Q1}")
print(f"Q3 : {Q3}")
print(f"IQR : {IQR}")

print(f"Limite inférieure : {limite_inf}")
print(f"Limite supérieure : {limite_sup}")


valeurs_aberrantes = clients[
    (clients["salaire"] < limite_inf) |
    (clients["salaire"] > limite_sup)
]

print("\nValeurs aberrantes détectées :")

display(valeurs_aberrantes)


print("\n===== Création de nouvelles variables =====")

clients["salaire_annuel"] = clients["salaire"] * 12

print("Colonne salaire_annuel créée.")

display(clients)


print("\n===== Préparation des données =====")

features = [
    "age",
    "salaire",
    "anciennete",
    "nombre_commandes",
    "depenses_totales"
]

X = clients.drop(columns="churn")

y = clients["churn"]


print(f"Dimension de X : {X.shape}")

print(f"Dimension de y : {y.shape}")


print("\nVérification du nombre de lignes :")

print(X.shape[0] == y.shape[0])


print("\nValeurs manquantes dans X :")

print(X.isna().sum().sum())


print("\nValeurs manquantes dans y :")

print(y.isna().sum())


print("\n===== Encodage des variables catégorielles =====")

X = pd.get_dummies(
    X,
    drop_first=True
)

print(f"Nouvelle dimension de X : {X.shape}")


print("\n===== Visualisation =====")


# Salaire moyen par profession

salaire_profession = (
    clients.groupby("profession")["salaire"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

plt.bar(
    salaire_profession.index,
    salaire_profession.values
)

plt.title("Salaire moyen par profession")

plt.xlabel("Profession")

plt.ylabel("Salaire moyen (MAD)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# Distribution des dépenses

plt.figure(figsize=(9, 5))

plt.hist(
    clients["depenses_totales"],
    bins=8
)

plt.title("Distribution des dépenses")

plt.xlabel("Dépenses totales (MAD)")

plt.ylabel("Nombre de clients")

plt.tight_layout()

plt.show()


print("\nVisualisations terminées.")

print("\n===== Analyse terminée =====")
```
